```python
import os
import pandas as pd
import sys

# --- Configuration ---
# Set the base directory containing your telemetry data.
TELEMETRY_BASE_DIR = 'Telemetry/Real'

def round_telemetry_files():
    """
    Finds all telemetry CSVs, rounds all numeric data to 3 decimal places,
    and overwrites the original files.
    """
    
    # --- DANGER ZONE: User Confirmation ---
    print("="*60)
    print("!!! WARNING: POTENTIALLY DESTRUCTIVE OPERATION !!!")
    print("="*60)
    print(f"This script will find all .csv files in the '{TELEMETRY_BASE_DIR}' directory")
    print("and its subfolders.")
    print("\nIt will PERMANENTLY OVERWRITE each file by rounding all of its")
    print("numeric data to 3 decimal places.")
    print("\nTHIS ACTION CANNOT BE UNDONE.")
    print("It is strongly recommended to back up your data first.")
    print("="*60)

    # Ask for explicit confirmation from the user
    response = input("Are you sure you want to continue? (Type 'yes' to proceed): ").lower()
    if response != 'yes':
        print("Aborting script. No files have been changed.")
        sys.exit() # Exits the script
    
    print("\nConfirmation received. Starting file processing...\n")

    # --- File Processing Loop ---
    if not os.path.isdir(TELEMETRY_BASE_DIR):
        print(f"Error: Base directory not found at '{TELEMETRY_BASE_DIR}'")
        return

    # Walk through all directories and files in the base directory
    files_processed = 0
    for root, _, files in os.walk(TELEMETRY_BASE_DIR):
        for file in files:
            if file.endswith('.csv'):
                file_path = os.path.join(root, file)
                
                try:
                    # Read the CSV file
                    df = pd.read_csv(file_path)

                    # Round all numeric columns in the DataFrame to 3 decimal places
                    # The .round() method automatically applies to all numeric columns
                    df = df.round(3)
                    
                    # Overwrite the original file
                    # index=False is crucial to avoid adding a new index column
                    df.to_csv(file_path, index=False)
                    
                    print(f"Processed and overwrote: {file_path}")
                    files_processed += 1

                except Exception as e:
                    print(f"Error processing file {file_path}: {e}")

    print(f"\nScript finished. Processed and overwrote {files_processed} files.")


if __name__ == '__main__':
    round_telemetry_files()
```
